```python
import cv2
 
video = cv2.VideoCapture("./assets/video.mp4")
 
while True:

    success, img = video.read()
 
    if not success:

        print("Video finished or file not found.")

        break
 
    image_resized = cv2.resize(img, (340, 480))
 
    cv2.imshow("AI Video", image_resized)
 
    key = cv2.waitKey(20) & 0xFF
 
    if key == ord('q') or key == ord('Q'):

        print("Video stopped")

        break
 
video.release()

cv2.destroyAllWindows()
 
```
