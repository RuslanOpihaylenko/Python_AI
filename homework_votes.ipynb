```python
from collections import Counter
votes = ["apple", "banana", "apple", "orange", "banana", "apple", "apple", "apple", "orange"]
counts = Counter(votes)

leader = counts.most_common(1)[0]

print("Фрукт-лідер:", leader[0])
print("Кількість голосів:", leader[1])

print("Фрукт\tКількість")
print("-" * 20)

for fruit, count in counts.items():
    print(f"{fruit}\t{count}")

top_3 = counts.most_common(3)

print("ТОП-3:")
for place, (fruit, count) in enumerate(top_3, start=1):
    print(f"{place}. {fruit} — {count}")
```

Кількість голосів: 5
