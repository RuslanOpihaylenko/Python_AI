```python
math = [80, 92, 75, 60, 88]
physics = [70, 65, 90, 85, 78]
english = [95, 100, 85, 90, 87]

subjects = {
    "math": math,
    "physics": physics,
    "english": english
}

results = {}

for subject, scores in subjects.items():
    results[subject] = {
        "average": sum(scores) / len(scores),
        "min": min(scores),
        "max": max(scores)
    }

# Предмет з найвищим середнім результатом
best_subject = max(
    results,
    key=lambda subject: results[subject]["average"]
)

print("Предмет з найвищим середнім результатом:", best_subject)

print(f"{'Предмет':<12} {'Середній':<12} {'Мінімум':<10} {'Максимум':<10}")
print("-" * 44)

for subject, data in results.items():
    print(
        f"{subject:<12} "
        f"{data['average']:<12.2f} "
        f"{data['min']:<10} "
        f"{data['max']:<10}"
    )
```
