```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
scores = np.random.normal(
    loc=70,     # середнє
    scale=10,   # стандартне відхилення
    size=50000   # кількість виборки
)

print("Обсяг генеральної сукупності:", len(scores)) 
print("Середнє генеральної сукупності:", round(np.mean(scores), 4)) 
print("Стандартне відхилення генеральної сукупності:", round(np.std(scores), 4))

def calculate_sample_means(scores, n, num_samples=1000): 
        sample_means = [] 
        for _ in range(num_samples): 
            sample = np.random.choice(scores, size=n, replace=True) 
            sample_means.append(np.mean(sample)) 
        return np.array(sample_means) 

num1 = 5 
num2 = 30 
num_samples = 1000 
means_n1 = calculate_sample_means(scores, num1, num_samples) 
means_n2 = calculate_sample_means(scores, num2, num_samples) 

fig, axes = plt.subplots(1, 2, figsize=(14, 5)) 
axes[0].hist(means_n1, bins=30, edgecolor="black")
axes[0].set_title(f"Розподіл вибіркових середніх (n = {num1})") 
axes[0].set_xlabel("Вибіркове середнє") 
axes[0].set_ylabel("Частота") 
axes[0].grid(axis="y", alpha=0.3) 

axes[1].hist(means_n2, bins=30, edgecolor="black") 
axes[1].set_title(f"Розподіл вибіркових середніх (n = {num2})")
axes[1].set_xlabel("Вибіркове середнє") 
axes[1].set_ylabel("Частота") 
axes[1].grid(axis="y", alpha=0.3) 
plt.tight_layout()
plt.show() 


print("\nРезультати моделювання:") 
print(f"\nn = {num1}")
print("Середнє вибіркових середніх:", round(np.mean(means_n1), 4))
print("Стандартне відхилення вибіркових середніх:", round(np.std(means_n1), 4))
print(f"\nn = {num2}") 

print("Середнє вибіркових середніх:", round(np.mean(means_n2), 4)) 
print("Стандартне відхилення вибіркових середніх:", round(np.std(means_n2), 4))
```
