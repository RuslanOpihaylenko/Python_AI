```python
import numpy as np

arr = np.random.randint(0, 51, 20)

print("Исходный массив:", arr)

threshold = int(input("Введите пороговое значение: "))

count = np.sum(arr > threshold)
print("Количество элементов, превышающих порог:", count)

max_value = np.max(arr)
max_position = np.argmax(arr)

print("Максимальное значение:", max_value)
print("Позиция первого появления максимума:", max_position)

sorted_arr = np.sort(arr)[::-1]
print("Массив по убыванию:", sorted_arr)
```

```python
import numpy as np

lower = int(input("Введите нижнюю границу диапазона: "))
upper = int(input("Введите верхнюю границу диапазона: "))

if lower > upper:
    print("Ошибка: нижняя граница больше верхней!")
else:
    matrix = np.random.randint(lower, upper + 1, (5, 5))

    print("\nИсходная матрица:")
    print(matrix)

    diagonal = np.diag(matrix)
    print("\nГлавная диагональ:", diagonal)

    diagonal_sum = np.sum(diagonal)
    print("Сумма элементов главной диагонали:", diagonal_sum)

    matrix[np.triu_indices(5, k=1)] = 0

    print("\nМатрица после обнуления элементов выше главной диагонали:")
    print(matrix)
```
